```python
###Load the trainedmodel, scaler pickle, onehot
import pickle
from tensorflow.keras.models import load_model
model= load_model('model.h5')


with open('one_hot_encoder_geo.pkl', 'rb') as file:
    one_hot_encoder_geo= pickle.load(file)

with open('label_encoder.pkl', 'rb') as file:
    label_encoder= pickle.load(file)

with open('Scaler.pkl', 'rb') as file:
    scaler= pickle.load(file)
```

```python
#Example input data
input_data= {
    'CreditScore':600,
    'Geography': 'France',	
    'Gender': 'Male',	
    'Age': 40,	
    'Tenure': 3	,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
import pandas as pd
encoded_geo = one_hot_encoder_geo.transform([[input_data['Geography']]]).toarray()
geo_encoded_df = pd.DataFrame(encoded_geo,columns=one_hot_encoder_geo.get_feature_names_out(['Geography']))
geo_encoded_df
```

```python
input_data=pd.DataFrame([input_data])
input_data
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()   

label_encoder.fit(['Male', 'Female'])  

input_data['Gender'] = label_encoder.transform(
    input_data['Gender']
)

input_data
```

```python
input_data=pd.concat([input_data.drop('Geography', axis=1),geo_encoded_df], axis=1)
input_data
```

```python
scaled_input= scaler.transform(input_data)
scaled_input
```

```python
prediction=model.predict(scaled_input)
prediction
```

```python
churn_proba= prediction[0][0]
print(churn_proba)
if churn_proba>0.5:
    print("The customer is likely to churn.")
else:
    print("The customer is unlikely to churn.")
```
